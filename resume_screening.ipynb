extracting text from resumes

```python
import os
import fitz

def extract_text_from_pdf(pdf_path):
    text = ""
    doc = fitz.open(pdf_path)
    for page in doc:
        text += page.get_text()
    doc.close()
    return text

resume_folder = "/content/resume_screening/resumes"
output_folder = "/content/resume_screening/output"

os.makedirs(output_folder, exist_ok=True)

for file in os.listdir(resume_folder):
    if file.endswith(".pdf"):
        text = extract_text_from_pdf(os.path.join(resume_folder, file))

        with open(os.path.join(output_folder, file.replace(".pdf", ".txt")), "w") as f:
            f.write(text)

print("Done")
```
